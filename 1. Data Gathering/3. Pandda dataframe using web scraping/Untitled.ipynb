```python
link = "https://www.pakpositions.com/jobs/stoneshed-jobs-in-islamabad-lead-verification-"

job_title = []
company = []
salary = []
job_type = []
location = []
post_date = []
last_date = []
experience = []
education = []
apply = []

for i in range(10500,10530):
    webpage = requests.get('https://www.pakpositions.com/jobs/stoneshed-jobs-in-islamabad-lead-verification-{}'.format(i),headers=headers).text
    soup=BeautifulSoup(webpage,'lxml')

    if len(webpage)<=11061:
        continue
    else:    
        job_title.append(soup.find('h1', class_="jobname").text)
        company.append(soup.find('a', class_="companyname").text)
        salary.append(soup.find('ul', class_="reqlist").text.strip().split("\n")[16])
        job_type.append(soup.find('ul', class_="reqlist").text.strip().split("\n")[11])
        location.append(soup.find('div', class_="location").text.strip().split('\xa0')[0])
        education.append(soup.find('ul', class_="reqlist").text.strip().split("\n")[26])
        post_date.append(soup.find('ul', class_="reqlist").text.strip().split("\n")[41])
        last_date.append(soup.find('ul', class_="reqlist").text.strip().split("\n")[36])
        experience.append(soup.find('ul', class_="reqlist").text.strip().split("\n")[31])
        apply.append(link+str(i))

d = {'Job_Title':job_title,'Company':company,'Issue_date':post_date,'Last_date':last_date,'Location':location,'Experience':experience,'Salary':salary,'Job_type':job_type,'Education':education,'Apply':apply}
df4 = pd.DataFrame(d)
```
