```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('ecommerce_dataset_10000.csv',usecols=[1,2,9,10,11,14,16,17])
```

```python
df.head()
```

# Feature Construction

```python
df['Full_Name'] = df['first_name'] + " " + df['last_name']
```

```python
df.drop(columns=['first_name','last_name'],inplace=True)
```

```python
df['order_status'].value_counts()
```

```python
df.head()
```

# Feature Splitting

```python
df['order_completion'] = df['order_status']
df['order_completion'].loc[df['order_status']=='Delivered'] = "Completed"
df['order_completion'].loc[df['order_status']!='Delivered'] = "Not Completed"
```
