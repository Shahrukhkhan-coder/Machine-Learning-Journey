```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("used_car_listings.csv")
```

```python
df.head()
```

# 1.Categorical Data

## a. Countplot

```python
sns.countplot(df["transmission"])
```

```python
df["transmission"].value_counts().plot(kind='bar')
```

## b. Pie chart

```python
df["condition"].value_counts().plot(kind='pie', autopct='%2f')
```

# 2. Numerical data

## a. Histogram

```python
plt.hist(df["mileage"], bins=4)
```

## b. Distplot

```python
sns.distplot(df["mileage"],bins=8)
```

## c. Boxplot

```python
sns.boxplot(df["price"])
```
