```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("restaurant_orders.csv")
```

```python
df.head()
```

# 1.Categorical Data

## a.Countplot

```python
sns.countplot(df["Food Item"])
```

```python
df["Food Item"].value_counts().plot(kind="bar")
```

## Pie Chart

```python
df["Payment Method"].value_counts().plot(kind="pie" , autopct="%.2f")
```

# 2.Numerical data

## a. Histogram

```python
import matplotlib.pyplot as plt
plt.hist(df["Price"],bins=5)
```

## b.Distplot

```python
sns.distplot(df["Price"])
```

## c. Boxplot

```python
sns.boxplot(df['Price'])
```
