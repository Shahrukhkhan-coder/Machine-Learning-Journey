```python
import pandas as pd
import seaborn as sns
```

```python
car=pd.read_csv('used_car_listings.csv')
```

```python
car.head()
```

# 1.ScatterPlot

```python
sns.scatterplot(x=car['mileage'],y=car['price'],hue=car['seller_type'],style=car['transmission'])
```

# 2. Bar Plot(N - C)

```python
sns.barplot(x=car['fuel_type'],y=car['price'],hue=car['seller_type'])
```

# 3. Distplot(N - C)

```python
sns.distplot(car[car['seller_type']=='Dealer']['mileage'],hist=False)
sns.distplot(car[car['seller_type']=='Certified Pre-Owned']['mileage'],hist=False)
sns.distplot(car[car['seller_type']=='Private Party']['mileage'],hist=False)
```

# 4. Heat Map(C - C)

```python
sns.heatmap(pd.crosstab(car['fuel_type'],car['condition']))
```

# 5. Pairplot

```python
sns.pairplot(car)
```

```python
new=car.groupby('year').sum()
```
