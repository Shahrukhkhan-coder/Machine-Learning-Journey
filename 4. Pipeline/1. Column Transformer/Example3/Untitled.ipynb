```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

```python
df = df.drop(columns=['PassengerId','Name','Ticket','Cabin'])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(df.drop(columns='Survived'),df['Survived'],test_size=0.2)

x_train.shape,y_train.shape
```

```python
df.isnull().sum()
```

```python
# Without Transformer
```

```python
# SimpleImputer -> Age & Embarak
from sklearn.impute import SimpleImputer

si = SimpleImputer()
si1 = SimpleImputer(strategy='most_frequent')

x_train_Age = si.fit_transform(x_train[['Age']])
x_test_Age = si.transform(x_test[['Age',]])

x_train_Embarked = si1.fit_transform(x_train[['Embarked']])
x_test_Embarked = si1.transform(x_test[['Embarked']])
x_train_Age.shape
x_train_Embarked.shape
```

```python
# StandardScaler -> Age & Fare
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler1 = StandardScaler()

x_train_Age = scaler.fit_transform(x_train_Age)
x_test_Age = si.transform(x_test_Age)

x_train_Fare = si1.fit_transform(x_train[['Fare']])
x_test_Fare = si1.transform(x_test[['Fare']])
x_train_Age.shape
x_train_Fare.shape
```

```python
# OneHotEncoder -> Sex & Embarked
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop='first',sparse_output=False)
ohe1 = OneHotEncoder(drop='first',sparse_output=False)

x_train_Sex = ohe.fit_transform(x_train[['Sex']])
x_test_Sex = ohe.transform(x_test[['Sex']])

x_train_Embarked = ohe1.fit_transform(x_train[['Embarked']])
x_test_Embarked = ohe1.transform(x_test[['Embarked']])
x_train_Sex.shape
x_train_Embarked.shape
```

```python
#Extract remaining features

x_train_rem = x_train.drop(['Age','Embarked','Fare','Sex'],axis=1)
x_test_rem = x_test.drop(['Age','Embarked','Fare','Sex'],axis=1)
```

```python
# Concatenate them back all
x_train_transformed = np.concatenate((x_train_Age,x_train_Fare,x_train_Sex,x_train_Embarked,x_train_rem),axis=1)
x_test_transformed = np.concatenate((x_test_Age,x_test_Fare,x_test_Sex,x_test_Embarked,x_test_rem),axis=1)
```
