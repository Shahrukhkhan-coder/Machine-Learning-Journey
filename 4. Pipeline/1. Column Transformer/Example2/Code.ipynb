```python
import numpy as np
import pandas as pd
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv('restaurant_orders.csv',usecols=[2,3,4,5])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(df.drop(columns='Quantity'),df['Quantity'],test_size=0.2)

x_train.shape,y_train.shape
```

# without Transformer

```python
# One Hot Encoder -> Food Item & Category
ohe = OneHotEncoder(drop='first',sparse_output=False)

x_train_Food_Item_Category = ohe.fit_transform(x_train[['Food Item','Category']])
x_test_Food_Item_Category = ohe.transform(x_test[['Food Item','Category']])

x_train_Food_Item_Category.shape
```

```python
# StandardScaler -> Price & Quantity
scaler = StandardScaler()

x_train_Price = scaler.fit_transform(x_train[['Price']])
x_test_Price = scaler.fit_transform(x_test[['Price']])

y_train_Quantity = scaler.fit_transform(y_train.values.reshape(-1,1))
y_test_Quantity = scaler.fit_transform(y_test.values.reshape(-1,1))
```

```python
x_train_transformed = np.concatenate((x_train_Food_Item_Category,x_train_Price,y_train_Quantity),axis=1)
x_train_transformed = np.concatenate((x_test_Food_Item_Category,x_test_Price,y_test_Quantity),axis=1)
x_train_transformed.shape
```

# With Transformer

```python
from sklearn.compose import ColumnTransformer

Transformer1 = ColumnTransformer([
    ('trf1',OneHotEncoder(drop='first',sparse_output=False),['Food Item','Category']),
    ('trf2',StandardScaler(),['Price']),
   
])

Transformer1.fit_transform(x_train).shape
Transformer1.transform(x_test).shape
```
