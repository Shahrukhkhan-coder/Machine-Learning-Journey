```python
import numpy as np
import pandas as pd
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df.head()
```

```python
 df['city'].unique()
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(df.drop(columns='has_covid'),df['has_covid'],test_size=0.2)

x_train.shape,y_train.shape
```

# Without Transformer

```python
# SimpleImputer -> fever
si = SimpleImputer()

x_train_fever = si.fit_transform(x_train[['fever']])
x_test_fever = si.transform(x_test[['fever']])

x_train_fever.shape
```

```python
# StandardScaler -> Age & fever
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train[['fever','age']])
x_test_scaled = scaler.transform(x_test[['fever','age']])
x_train_scaled.shape
```

```python
# OrdinalEncoder -> cough
oe = OrdinalEncoder(categories=[['Mild','Strong']])

x_train_cough = oe.fit_transform(x_train[['cough']])
x_test_cough = oe.transform(x_test[['cough']])
x_train_cough.shape
```

```python
# OneHotEncoder -> gender & city
ohe = OneHotEncoder(drop='first',sparse_output=False)

x_train_gender_city = ohe.fit_transform(x_train[['gender','city']])
x_test_gender_city = ohe.transform(x_test[['gender','city']])
x_train_gender_city.shape
```

```python
x_train_transformed = np.concatenate((x_train_fever,x_train_scaled,x_train_scaled,x_train_gender_city),axis=1)
x_test_transformed = np.concatenate((x_test_fever,x_test_scaled,x_test_scaled,x_test_gender_city),axis=1)
x_train_transformed.shape
```

# With Transformer

```python
from sklearn.compose import ColumnTransformer

transformer = ColumnTransformer(transformers=[
    ('trf1',SimpleImputer(),['fever']),
    ('trf2',StandardScaler(),['fever','age']),
    ('trf3',OrdinalEncoder(categories='auto'),['cough']),
    ('trf4',OneHotEncoder(drop='first',sparse_output=False),['gender','city'])
])

transformer.fit_transform(x_train).shape
transformer.transform(x_test).shape
```
