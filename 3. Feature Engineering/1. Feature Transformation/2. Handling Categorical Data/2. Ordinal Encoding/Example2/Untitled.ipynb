```python
df = pd.read_csv('ecommerce_dataset_10000.csv')
```

```python
df.head()
```

```python
df = df[['age_group','category','quantity','unit_price','rating','review_text']]
```

```python
df.head()
```

```python
transformed_df = df[['category','quantity','unit_price','rating']]
```

```python
transformed_df.sample(5)
```

```python
test_df = df[['age_group','review_text']]
```

```python
test_df.head()
```

```python
test_df['review_text'].unique()
```

# Ordinal Encoder

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(categories=[['Teenagers','Adults','Senior'],['very bad','bad','average','good','very good']])

oe.fit(test_df)

test_df_transformed = oe.transform(test_df)
```

```python
test_df_transformed
```

```python
test_df_transformed.shape
```

```python
test_df_transformed = pd.DataFrame(test_df_transformed, columns=test_df.columns)
```

```python
test_df_transformed.head()
```

# Concatenating back Ordinal Encoder transformed to Original Dataframe

```python
transformed_df = pd.concat([transformed_df,test_df_transformed],axis=1)
```
