# Train Test Split

```python
x = df.iloc[:,0:4]
x
```

```python
y = df.iloc[:,-1]
y
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y,test_size=0.2,random_state=0)
```

# One Hot Encoding

```python
from sklearn.preprocessing import OneHotEncoder
```

Sparsity parameter used to avoid sparse matrix and at the end .toarray require to display result

dtypes parameter convert float ponit to int

```python
ohe = OneHotEncoder(drop='first',sparse_output=False,dtype=np.int32)
```

```python
x_train_new = ohe.fit_transform(x_train[['fuel','owner']])
x_test_new = ohe.transform(x_test[['fuel','owner']])
```

```python
x_train_new.shape
```

To concatenate One hot encoder columns and remaining columns

```python
np.hstack((x_train[['brand','km_driven']].values,x_train_new))
np.hstack((x_test[['brand','km_driven']].values,x_test_new))
```
