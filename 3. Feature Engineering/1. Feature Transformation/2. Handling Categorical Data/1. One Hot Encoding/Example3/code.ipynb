# Train test split

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(df.drop(columns='income'), df['income'],test_size=0.2,random_state=0)
x_train.shape, x_test.shape
```

# One Hot Encoder

```python
counts1 = df['marital.status'].value_counts()
counts2 = df['occupation'].value_counts()
counts =pd.concat((counts1, counts2))
threshold=1000
repl=counts[counts<=threshold].index
pd.get_dummies(x_train[['marital.status','occupation']].replace(repl,'uncommnon'))
#pd.get_dummies(x_train[['occupation']].replace(repl,'uncommnon'))
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop='first',sparse_output=False,dtype=np.int32)

x_train_new = ohe.fit_transform(x_train[['marital.status','occupation','relationship','sex','race']].replace(repl,'other'))
x_test_new = ohe.fit_transform(x_test[['marital.status','occupation','relationship','sex','race']].replace(repl,'other'))
```
