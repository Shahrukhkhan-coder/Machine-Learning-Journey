# 
Train test split

```python
x = df[['make','model','body_type','fuel_type','transmission','mileage','price','condition','seller_type']]
```

```python
y = df['price']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y,test_size=0.2,random_state=0)
```

# One Hot Encoding

```python
counts = df['model'].value_counts()
```

```python
df['model'].value_counts()
threshold=40
repl = counts[counts<=threshold].index
#pd.get_dummies(x_train['model'].replace(repl,'uncommon'))
```

```python

from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(drop='first',sparse_output=False,dtype=np.int32)

x_train_new = ohe.fit_transform(x_train[['make','body_type','fuel_type','transmission','seller_type','model']].replace(repl,'uncommon'))
x_test_new = ohe.transform(x_test[['make','body_type','fuel_type','transmission','seller_type','model']].replace(repl,'uncommon'))
```
