```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, KFold
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report
from catboost import CatBoostClassifier
```

```python
# Create sample dataset
np.random.seed(42)
n = 1000
df = pd.DataFrame({
    "city": np.random.choice(["Islamabad", "Lahore", "Karachi", "Peshawar", "Quetta"],size=n),
    "browser": np.random.choice(["Chrome", "Firefox", "Safari", "Edge"],size=n),
    "product": np.random.choice(["Laptop", "Phone", "Tablet", "Monitor", "Keyboard", "Mouse"],size=n),
    "customer_id": np.random.choice([f"C{i}" for i in range(1, 101)],size=n),
    "age": np.random.randint(18, 60, size=n),
    "income": np.random.randint(30000, 200000, size=n)})

# Create binary target
df["purchased"] = ((df["income"] > 100000).astype(int) | (df["age"] > 45).astype(int))

# Convert target to 0/1
df["purchased"] = (df["purchased"] > 0).astype(int)

df.head()
```

```python
X = df.drop("purchased", axis=1)
y = df["purchased"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
model = CatBoostClassifier(iterations=500, depth=6, learning_rate=0.05, verbose=False)
```

```python
# Train

model.fit(X_train, y_train, cat_features=["city", "product", "browser", "customer_id"])
```

```python
# Predict

predictions = model.predict(X_test)
```

```python
# Evaluate

accuracy = accuracy_score(y_test, predictions)
print("Target-Mean Accuracy:", accuracy)
```
