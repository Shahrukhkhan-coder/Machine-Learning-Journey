# Working With Date

```python
# Converting to datetime datatype
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

```python
# Extract Year
date['date_year'] = date['date'].dt.year
```

```python
date.sample(5)
```

```python
# Extract Month
date['date_month_num'] = date['date'].dt.month

date.sample(5)
```

```python
# Extract Month
date['date_month_name'] = date['date'].dt.month_name()

date.sample(5)
```

```python
# Extract Day
date['date_day'] = date['date'].dt.day

#date = date.drop(columns=['product_id','city_id','date_month'])
date.sample(5)
```

```python
# Extract Day_of_week
date['date_dayofweek'] = date['date'].dt.dayofweek

date.sample(5)
```

```python
# Extract Day_of_week -> name
date['date_dow_name'] = date['date'].dt.day_name()

date.sample(5)
```

```python
# Is weekened?

date['day_is_weekened'] = np.where(date['date_dow_name'].isin(['Saturday','Sunday']),1,0)

date.sample(3)
```

```python
# Extract week of year 
date['date_week'] = date['date'].dt.isocalendar().week

date.sample(5)
```

```python
# Extract Quarter
date['date_quarter'] = date['date'].dt.quarter

date.sample(3)
```

```python
# Extract Semester
date['date_semerter'] = np.where(date['date_quarter'].isin([1,2]),1,2)

date.sample(3)
```

# Extract time elapse between dates

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - date['date']
```

```python
year_passed = (today - date['date']).dt.days
year_passed
```

```python
month_passed = ((today.year - date['date'].dt.year) * 12 + today.month - date['date'].dt.month)
month_passed
```

# Working With Time

```python
# Convert to datetime Format

time['date'] = pd.to_datetime(time['date'])
```

```python
time.info()
```

```python
# Extract time part

time['time'] = time['date'].dt.time

time.sample(3)
```

```python
time['hours'] = time['date'].dt.hour
time['minutes'] = time['date'].dt.minute
time['seconds'] = time['date'].dt.second

time.sample(3)
```

```python
# Time Difference

today - time['date']
```

```python
# in seconds

(today - time['date'])/np.timedelta64(1,'s')
```

```python
# In minutes

(today - time['date'])/np.timedelta64(1,'m')
```

```python
# In hours

(today - time['date'])/np.timedelta64(1,'h')
```
