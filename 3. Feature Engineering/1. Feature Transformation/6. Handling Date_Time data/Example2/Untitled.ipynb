```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('public_transport_delays.csv',usecols=['date','time','delayed'])
```

```python
df.head()
```

```python
df.info()
```

```python
# Coverting to datetime datatype
df['date'] = pd.to_datetime(df['date'])
df['time'] = pd.to_datetime(df['time'])
```

```python
df.info()
```

```python
# Working With date
```

```python
df['date_year'] = df['date'].dt.year
df.head()
```

```python
df['date_month_num'] = df['date'].dt.month
df.head()
```

```python
df['date_month_name'] = df['date'].dt.month_name()
df.head()
```

```python
df['date_day'] = df['date'].dt.day
df.head()
```

```python
df['date_day_name'] = df['date'].dt.day_name()
df.head()
```

```python
df['date_dayofweek'] = df['date'].dt.dayofweek
df.head()
```

```python
df['date_is_weekened'] = np.where(df['date_dayofweek'].isin(['Saturady','Sunday']),0,1)
df.head()
```

```python
df['date_week'] = df['date'].dt.isocalendar().week
df.head()
```

```python
df['date_quarter'] = df['date'].dt.quarter
df.head()
```

```python
df['date_semester'] = np.where(df['date_quarter'].isin([1,2]),1,2)
df.head()
```

```python
# time elapse between date
```

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - df['date']
```

```python
day_passed = (today - df['date']).dt.days
day_passed
```

```python
month_passed = ((today.year - df['date'].dt.year) * 12 + today.month - df['date'].dt.month)
month_passed
```

```python
year_passed = (today.year - df['date'].dt.year)
year_passed
```

```python
# Working with Time
```

```python
df['time_part'] = df['time'].dt.time
df.sample(3)
```

```python
df['hour'] = df['time'].dt.hour
df['minute'] = df['time'].dt.minute
df['second'] = df['time'].dt.second

df.sample(3)
```

```python
# Time Difference

today - df['time']
```

```python
# In second

(today - df['time'])/np.timedelta64(1,'s')
```

```python
# In minute

(today - df['time'])/np.timedelta64(1,'m')
```

```python
# In hour

(today - df['time'])/np.timedelta64(1,'h')
```
