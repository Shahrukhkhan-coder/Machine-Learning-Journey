```python
import  numpy as np 
import pandas as pd
```

```python
df = pd.read_csv('used_car_listings.csv',usecols=['mileage','price','condition','location','features'])
```

```python
df.head()
```

```python
df[['City','Province','Country']] = df['location'].str.split(',',expand=True)
```

```python
df['features'] = df['features'].fillna('unknown')
```

```python
features_split = df['features'].str.split(',',expand=True)
```

```python
unique_features = set(df['features'].str.split(',').explode().str.strip())
```

```python
for feature in unique_features:
    df[feature.strip()] = df['features'].apply(lambda x: feature.strip() in [f.strip() for f in x.split(',')])
```
