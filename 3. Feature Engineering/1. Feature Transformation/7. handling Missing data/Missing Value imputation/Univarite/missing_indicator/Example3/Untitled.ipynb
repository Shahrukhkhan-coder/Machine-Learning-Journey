```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.impute import MissingIndicator,SimpleImputer
```

```python
df = pd.read_csv('used_car_listings.csv',usecols=['trim','mileage','price','condition'])
```

```python
df.head()
```

```python
mi = MissingIndicator()

mi.fit(df)

df_missing = mi.transform(df)
```

```python
df_missing
```

```python
df['experience_na'] = df_missing[:,0] 

df['training_hours_na'] = df_missing[:,-1] 
```

```python
df.head()
```

```python
si = SimpleImputer(add_indicator=True,strategy='most_frequent')

si.fit(df)
```

```python
df = si.transform(df)
```
