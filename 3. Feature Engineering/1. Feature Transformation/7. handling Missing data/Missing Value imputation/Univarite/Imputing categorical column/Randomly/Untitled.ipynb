```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('house-train.csv',usecols=['GarageQual','FireplaceQu', 'SalePrice'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
x = df.drop(columns='SalePrice')
y = df['SalePrice']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
x_train
```

```python
for col in x_train.columns:
    np.random.seed(42)
    x_train[f'{col}_imputed'] = x_train[col]
    non_missing_values = x_train[col].dropna().values    
    missing_indices = x_train[col].isnull()
    x_train.loc[missing_indices, f'{col}_imputed'] = np.random.choice(non_missing_values, size=missing_indices.sum())
```
