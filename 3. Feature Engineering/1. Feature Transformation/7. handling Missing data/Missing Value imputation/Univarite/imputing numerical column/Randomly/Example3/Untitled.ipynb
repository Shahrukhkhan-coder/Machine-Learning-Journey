```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df  = pd.read_csv('data_science_job.csv',usecols=[2,8,11,12]) 
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
x = df.drop(columns='target')
y = df['target']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
for col in x_train.columns:
    np.random.seed(42)
    non_missing_values = x_train[col].dropna().values
    missing_indices = x_train[col].isnull()
    x_train[f'{col}_imputed'] = x_train[col]
    x_train.loc[missing_indices, f'{col}_imputed'] = np.random.choice(non_missing_values, size=missing_indices.sum())
```

```python
x_train
```

```python
sns.distplot(x_train['city_development_index'],label='Original',hist=False)
sns.distplot(x_train['city_development_index_imputed'],label = 'Imputed',hist=False)

plt.legend()
plt.show()
```

```python
sns.distplot(x_train['experience'],label='Original',hist=False)
sns.distplot(x_train['experience_imputed'],label = 'Imputed',hist=False)

plt.legend()
plt.show()
```

```python
sns.distplot(x_train['training_hours'],label='Original',hist=False)
sns.distplot(x_train['training_hours_imputed'],label = 'Imputed',hist=False)

plt.legend()
plt.show()
```

```python
print('Original variable variance: ', x_train['city_development_index'].var())
print('Variance after random imputation: ', x_train['city_development_index_imputed'].var())
print("")
print('Original variable variance: ', x_train['experience'].var())
print('Variance after random imputation: ', x_train['experience_imputed'].var())
print("")
print('Original variable variance: ', x_train['training_hours'].var())
print('Variance after random imputation: ', x_train['training_hours_imputed'].var())
```
