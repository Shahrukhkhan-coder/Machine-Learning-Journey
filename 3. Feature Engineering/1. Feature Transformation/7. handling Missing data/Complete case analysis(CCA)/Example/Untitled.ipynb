```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data_science_job.csv')
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
cols = [var for var in df.columns if df[var].isnull().mean()*100 < 5 and df[var].isnull().mean()*100 > 0]
cols
```

```python
df[cols].sample(3)
```

```python
len(df[cols].dropna()) / len(df)
```

```python
new_df = df[cols].dropna()
df.shape, new_df.shape
```

```python
fig = plt.figure()
ax = plt.subplot(111)

#original Data
df['training_hours'].hist(bins=50, ax=ax, color='red', density='True')

# data after cca, the argument alpha makes the color transparent, so we can
# see the overlay of the 2 distributions
new_df['training_hours'].hist(bins=50, ax=ax, color='green', density='True')
```

```python
fig = plt.figure()
ax = plt.subplot(111)

#original Data
df['training_hours'].plot.density(color='red')

# data after cca
new_df['training_hours'].plot.density(color='green')
```
