```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('used_car_listings.csv')
```

```python
df.head()
```

```python
df.isnull().mean()*100
```

```python
cols = [var for var in df.columns if df[var].isnull().mean()*100 < 10 and df[var].isnull().mean()*100 > 0]
cols
```

```python
len(df[cols].dropna()) / len(df)
```

```python
new_df = df[cols].dropna()
df.shape, new_df.shape
```

```python
temp = pd.concat([
    df['condition'].value_counts() / len(df),
    new_df['condition'].value_counts() / len(new_df)
],axis=1)

temp.columns= ['original','CCA']

temp
```

```python
temp = pd.concat([
    df['features'].value_counts() / len(df),
    new_df['features'].value_counts() / len(new_df)
],axis=1)

temp.columns= ['original','CCA']

temp
```
