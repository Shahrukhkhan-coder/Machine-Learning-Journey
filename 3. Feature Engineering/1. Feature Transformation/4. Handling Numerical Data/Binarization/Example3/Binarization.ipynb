```python
import numpy as np
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import accuracy_score

from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('ecommerce_dataset_10000.csv',usecols=['quantity','unit_price','rating'])
```

```python
df.head()
```

```python
x = df.iloc[:,[0,1]]
y = df.iloc[:,-1]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
y_train.head()
```

```python
# Without Binarization
```

```python
clf = DecisionTreeClassifier()

clf.fit(x_train,y_train)

y_pred = clf.predict(x_test)

accuracy_score(y_test,y_pred)
```

```python
np.mean(cross_val_score(clf,x,y,scoring='accuracy'))
```

```python
# Applying Binarization
```

```python
y_train = pd.DataFrame(y_train,columns=['rating'])
y_test = pd.DataFrame(y_test,columns=['rating'])
y_train.head()
```

```python
from sklearn.preprocessing import Binarizer

trf = ColumnTransformer([
    ('trf',Binarizer(copy=False,threshold=2),['rating'])
],remainder='passthrough')

y_train_trf = trf.fit_transform(y_train)
y_test_trf = trf.transform(y_test)

pd.DataFrame(y_train_trf,columns=['rating']).head()
```

```python
# With Binarization
```

```python
clf = DecisionTreeClassifier()

clf.fit(x_train,y_train_trf)

y_pred2 = clf.predict(x_test)

accuracy_score(y_test,y_pred2)
```

```python
y = pd.DataFrame(y,columns=['rating'])
y_trf = trf.transform(y)
np.mean(cross_val_score(clf,x,y_trf,scoring='accuracy'))
```
