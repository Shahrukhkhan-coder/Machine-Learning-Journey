```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import scipy.stats as stats
```

```python
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score

from sklearn.linear_model import LinearRegression

from sklearn.metrics import r2_score

from sklearn.preprocessing import PowerTransformer
```

```python
df = pd.read_csv('train.csv',usecols=[1,5,9])
```

```python
df.head()
```

```python
df['Age'].fillna(df['Age'].mean(),inplace=True)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
x = df.iloc[:,1:3]
y = df.iloc[:,-1]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42) 
```

```python
# Model training Before Power_Transformer
```

```python
lr = LinearRegression()

lr.fit(x_train ,y_train)

y_pred = lr.predict(x_test)

r2_score(y_test,y_pred)
```

```python
np.mean(cross_val_score(lr, x, y, scoring="r2"))
```

```python
# Power_Transformer[box_cox]
```

```python
pt = PowerTransformer(method='box-cox')

x_train_transformed = pt.fit_transform(x_train+0.000000001)
x_test_transformed = pt.fit_transform(x_test+0.000000001)

pd.DataFrame({'col':x_train.columns,'Lambdas':pt.lambdas_})
```

```python
# model training After Power_Transformation[box-cox]
```

```python
lr = LinearRegression()

lr.fit(x_train_transformed ,y_train)

y_pred2 = lr.predict(x_test_transformed)

r2_score(y_test,y_pred2)
```

```python
x_transformed = pt.fit_transform(x+0.00000000001)
np.mean(cross_val_score(lr, x_transformed, y, scoring="r2"))
```

```python
# Power_Transformer[Yeo-Johnson]
```

```python
pt1 = PowerTransformer()

x_train_transformed2 = pt1.fit_transform(x_train)
x_test_transformed2 = pt1.fit_transform(x_test)

pd.DataFrame({'col':x_train.columns,'Lambdas':pt1.lambdas_})
```

```python
# Model Training After Power_Transformation[Yeo-JohnSon]
```

```python
lr = LinearRegression()

lr.fit(x_train_transformed2 ,y_train)

y_pred3 = lr.predict(x_test_transformed2)

r2_score(y_test,y_pred3)
```

```python
x_transformed2 = pt1.fit_transform(x)
np.mean(cross_val_score(lr, x_transformed2, y, scoring="r2"))
```

```python
# Side by Side Comparsion of boc-cox & Yeo-Johnson Lambdas
pd.DataFrame({'col':x_train.columns,'box-cox_lambda':pt.lambdas_, 'Yeo-johnson_Lambdas':pt1.lambdas_})
```
