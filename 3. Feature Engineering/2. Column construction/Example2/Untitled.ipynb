```python
import numpy as np 
import pandas as pd
```

```python
df = pd.read_csv('bmw.csv')
```

```python
df.head()
```

# feature Construction

```python
df['Grand_Total'] = df['price'] + df['tax']
```

```python
df.head()
```

# Feature Splitting

```python
df['Car_Type'] = df['mileage']
df['Car_Type'].loc[df['mileage'] <= 1000] = 'Not Too Old'
df['Car_Type'].loc[df['mileage'] > 1000] = 'Old'
df['Car_Type'].loc[df['mileage'] <= 10] = 'New'
```
