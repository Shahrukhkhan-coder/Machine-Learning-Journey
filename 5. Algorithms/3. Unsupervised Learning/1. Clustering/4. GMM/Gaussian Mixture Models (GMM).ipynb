```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture
```

```python
# Create data
X, y_true = make_blobs(
    n_samples=500,
    centers=3,
    cluster_std=[1.0, 2.0, 0.8],
    random_state=42)
```

```python
# Train GMM
gmm = GaussianMixture(
    n_components=3,
    covariance_type="full",  #full, tied, diag, spherical
    random_state=42)
```

```python
gmm.fit(X)
```

```python
# Hard cluster labels
labels = gmm.predict(X)
```

```python
# Probabilities or soft cluster labels
probabilities = gmm.predict_proba(X)
```

```python
print("First 5 probability predictions:")
print(probabilities[:5])
```

```python
# Visualization
plt.figure(figsize=(8, 6))

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels)

plt.title("Gaussian Mixture Model")

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.show()
```

```python
# Prediction for a New Point
new_point = np.array([[2.0, 3.0]])

cluster = gmm.predict(new_point)

probability = gmm.predict_proba(new_point)

print("Predicted cluster:", cluster)

print("Cluster probabilities:", probability)
```
