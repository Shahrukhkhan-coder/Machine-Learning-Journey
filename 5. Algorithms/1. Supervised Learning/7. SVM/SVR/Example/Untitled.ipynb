```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVR
```

```python
# Create sample data
X = np.sort(5 * np.random.rand(100, 1), axis=0)
y = np.sin(X).ravel()
```

## Key Parameters 
- kernel: Type of SVM (e.g., 'linear', 'rbf', 'poly')
- C: Regularization parameter (higher = less tolerance for error)
- gamma: Defines influence of a single training example (for RBF/poly)
- epsilon: Margin of tolerance where no penalty is given

```python
# Create SVR models with different kernels
svr_rbf = SVR(kernel='rbf', C=100, gamma=0.1, epsilon=0.1)
svr_linear = SVR(kernel='linear', C=100, epsilon=0.1)
svr_poly = SVR(kernel='poly', C=100, degree=3, epsilon=0.1)
```

```python
# Fit models
svr_rbf.fit(X, y)
svr_linear.fit(X, y)
svr_poly.fit(X, y)
```

```python
# Predictions
y_rbf = svr_rbf.predict(X)
y_linear = svr_linear.predict(X)
y_poly = svr_poly.predict(X)
```

```python
# Plot results
plt.scatter(X, y, color='darkorange', label='data')
plt.plot(X, y_rbf, color='navy', label='RBF model')
plt.plot(X, y_linear, color='red', label='Linear model')
plt.plot(X, y_poly, color='blue', label='Polynomial model')

plt.xlabel('X')
plt.ylabel('y')
plt.title('Support Vector Regression')
plt.legend()
plt.show()
```

```python
from sklearn.metrics import r2_score

rbf = r2_score(y,y_rbf)
linear = r2_score(y,y_linear)
poly = r2_score(y,y_poly)

print("The RBF Kernel result:",rbf)
print("The Linear Kernel result:",linear)
print("The Poly Kernel result:",poly)
```
