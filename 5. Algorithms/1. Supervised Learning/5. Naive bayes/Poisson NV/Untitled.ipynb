```python
import numpy as np
import math
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
class PoissonNB:
    def fit(self, X, y):
        self.classes = np.unique(y)
        self.lambda_ = {}
        self.priors = {}

        for c in self.classes:
            X_c = X[y == c]
            self.lambda_[c] = np.mean(X_c, axis=0)
            self.priors[c] = X_c.shape[0] / X.shape[0]

    def poisson_pmf(self, class_idx, x):
        lam = self.lambda_[class_idx]
        return (np.exp(-lam) * (lam ** x)) / np.maximum(1, np.vectorize(math.factorial)(x))

    def predict(self, X):
        predictions = []
        for x in X:
            posteriors = []
            for c in self.classes:
                prior = np.log(self.priors[c])
                likelihood = np.sum(np.log(self.poisson_pmf(c, x) + 1e-9))
                posteriors.append(prior + likelihood)
            predictions.append(self.classes[np.argmax(posteriors)])
        return np.array(predictions)
```

```python
# Example count dataset (real-like)
X = np.random.poisson(lam=3, size=(200, 4))  # count data
y = np.random.randint(0, 2, size=200)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
model = PoissonNB()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print("Poisson NB Accuracy:", accuracy_score(y_test, y_pred))
```

## Poisson NB

Use it when your data:

- Contains count values (0, 1, 2, 3, …)
- Represents rate of occurrence

Examples:
- Number of emails received per hour
- Number of website clicks
- Count of rare events
- Word counts (sometimes, though Multinomial is more common)

## Simple Intuition

Think like this:

- “Is my data numeric?” → Gaussian
- “Am I counting things?” → Multinomial
- “Is it just yes/no?” → Bernoulli
- “Am I counting how often something happens in a fixed interval?” -> Poisson
