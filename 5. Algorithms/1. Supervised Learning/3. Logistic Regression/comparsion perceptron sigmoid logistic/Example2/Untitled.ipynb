```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('heart.csv').drop(columns=['Sex','ChestPainType','RestingECG','ST_Slope','ExerciseAngina'])
```

```python
df.head()
```

```python
x = df.drop(columns=['HeartDisease'])
y = df['HeartDisease']
```

```python
plt.figure(figsize=(10,6))
plt.scatter(y==0,y==1,c=y,cmap='winter')
```

```python
def perceptron(x,y):

    x = np.insert(x,0,1,axis=1)
    weights = np.ones(x.shape[1])
    lr = 0.1

    for i in range(1000):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(x[j],weights))
        weights = weights - lr * (y[j]-y_hat) * x[j] 

    return weights[0],weights[1:]
```

```python
def step(z):
    return 1 if z>0 else 0
```

```python
intercept_, coef_ = perceptron(x,y)
```

```python
intercept_, coef_
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(y==0,y==1,c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
