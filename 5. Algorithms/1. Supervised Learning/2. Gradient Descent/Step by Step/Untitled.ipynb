```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_regression
```

```python
x, y = make_regression(n_samples=4, n_features=1, n_targets=1, noise=80, random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
reg.fit(x,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
y_pred = (78.35 * x) + 100
```

```python
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label='OLS')
plt.plot(x,y_pred,color='green',label='b=0')
plt.legend()
```

```python
m = 78.35
b = 100

loss_slope = -2 * np.sum(y - m*x.ravel() - b) 
loss_slope
```

```python
lr = 0.1

loss_size = loss_slope*lr
loss_size
```

```python
b = b - loss_size
b
```

```python
y_pred1 = (78.35 * x) + b

plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label='OLS')
plt.plot(x,y_pred,color='green',label='b=100')
plt.plot(x,y_pred1,color='blue',label=b)
plt.legend()
```

```python
# Iter 2
loss_slope = -2 * np.sum(y - m*x.ravel() - b) 
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred2 = (78.35 * x) + b

plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label='OLS')
plt.plot(x,y_pred,color='green',label='b=100')
plt.plot(x,y_pred1,color='blue',label=b)
plt.plot(x,y_pred2,color='pink',label=b)
plt.legend()
```

```python
#Iter 3
loss_slope = -2 * np.sum(y - m*x.ravel() - b) 
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred3 = (78.35 * x) + b

plt.figure(figsize=(15,15))
plt.scatter(x,y)
plt.plot(x,reg.predict(x),color='red',label='OLS')
plt.plot(x,y_pred,color='green',label='b=100')
plt.plot(x,y_pred1,color='blue',label=b)
plt.plot(x,y_pred2,color='pink',label=b)
plt.plot(x,y_pred3,color='purple',label=b)
plt.legend()
```

```python
b = -100
m = 78.35
lr = 0.01

epochs = 50

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*x.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * x + b

  plt.plot(x,y_pred)

plt.scatter(x,y)
```
