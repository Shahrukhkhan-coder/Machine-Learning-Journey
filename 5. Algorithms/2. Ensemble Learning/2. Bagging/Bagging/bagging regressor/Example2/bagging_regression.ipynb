```python
import numpy as np
import pandas as pd
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv('advertising.csv')
```

```python
df.head()
```

```python
X = df.iloc[:,0:3]
Y = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X, Y , train_size=0.80, test_size=0.20, random_state=123)
print('Train/Test Sets Sizes : ',X_train.shape, X_test.shape, Y_train.shape, Y_test.shape)
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
knn = KNeighborsRegressor()
```

```python
lr.fit(X_train,Y_train)
dt.fit(X_train,Y_train)
knn.fit(X_train,Y_train)
```

```python
y_pred1 = lr.predict(X_test)
y_pred2 = dt.predict(X_test)
y_pred3 = knn.predict(X_test)
```

```python
print("R^2 score for LR",r2_score(Y_test,y_pred1))
print("R^2 score for DT",r2_score(Y_test,y_pred2))
print("R^2 score for KNN",r2_score(Y_test,y_pred3))
```

## Bagging using Linear Regression

```python
from sklearn.ensemble import BaggingRegressor

bag_regressor = BaggingRegressor(estimator=LinearRegression(),random_state=1)
bag_regressor.fit(X_train, Y_train)
```

```python
Y_preds = bag_regressor.predict(X_test)
r2_score(Y_test, Y_preds)
```

```python
## Using Decision Tree
```

```python
from sklearn.ensemble import BaggingRegressor

bag_regressor = BaggingRegressor(estimator=DecisionTreeRegressor(),random_state=1)
bag_regressor.fit(X_train, Y_train)
```

```python
Y_preds = bag_regressor.predict(X_test)
r2_score(Y_test, Y_preds)
```
