```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler

from sklearn.svm import OneClassSVM
```

### CREATE NORMAL DATA

```python
X_normal, _ = make_blobs(
    n_samples=1000,
    centers=1,
    cluster_std=1.0,
    random_state=42)
```

### CREATE ARTIFICIAL ANOMALIES

```python
rng = np.random.RandomState(42)

X_anomaly = rng.uniform(
    low=-8,
    high=8,
    size=(50, 2))
```

###  COMBINE DATA

```python
X = np.vstack([
    X_normal,
    X_anomaly])
```

### SCALE

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

### CREATE MODEL

```python
model = OneClassSVM(
    kernel="rbf",
    nu=0.05,
    gamma="scale")
```

### TRAIN

```python
model.fit(X_scaled)
```

### PREDICT

```python
predictions = model.predict(X_scaled)
```

### ANOMALY SCORE

```python
scores = -model.decision_function(X_scaled)
```

### VISUALIZATION

```python
normal = predictions == 1
anomaly = predictions == -1

plt.figure(figsize=(10, 7))

plt.scatter(
    X[normal, 0],
    X[normal, 1],
    label="Normal")

plt.scatter(
    X[anomaly, 0],
    X[anomaly, 1],
    label="Anomaly")

plt.title("One-Class SVM Anomaly Detection")

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.legend()

plt.show()
```
