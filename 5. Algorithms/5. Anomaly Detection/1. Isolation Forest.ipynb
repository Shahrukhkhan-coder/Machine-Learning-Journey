```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.ensemble import IsolationForest
```

### CREATE NORMAL DATA

```python
X_normal, _ = make_blobs(
    n_samples=1000,
    centers=1,
    cluster_std=1.0,
    random_state=42
)
```

### CREATE ARTIFICIAL ANOMALIES

```python
rng = np.random.RandomState(42)

X_anomaly = rng.uniform(
    low=-8,
    high=8,
    size=(50, 2))
```

###  COMBINE DATA

```python
X = np.vstack([
    X_normal,
    X_anomaly])


y_true = np.concatenate([
    np.ones(len(X_normal)),
    -np.ones(len(X_anomaly))])
```

### CREATE MODEL

```python
model = IsolationForest(
    n_estimators=200,
    contamination=0.05,
    random_state=42)
```

### TRAIN

```python
model.fit(X)
```

### PREDICT

```python
y_pred = model.predict(X)
```

### ANOMALY SCORE

```python
anomaly_score = -model.score_samples(X)
```

### CREATE RESULT DATAFRAME

```python
result = pd.DataFrame({
    "feature_1": X[:, 0],
    "feature_2": X[:, 1],
    "prediction": y_pred,
    "anomaly_score": anomaly_score,
    "actual": y_true})
```

### DISPLAY MOST ANOMALOUS POINTS

```python
print(result.sort_values("anomaly_score",ascending=False).head(20))
```

### VISUALIZATION

```python
plt.figure(figsize=(10, 7))

normal = result[result["prediction"] == 1]
anomalies = result[result["prediction"] == -1]

plt.scatter(
    normal["feature_1"],
    normal["feature_2"],
    label="Normal")

plt.scatter(
    anomalies["feature_1"],
    anomalies["feature_2"],
    label="Anomaly")

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.title("Isolation Forest Anomaly Detection")

plt.legend()

plt.show()
```
