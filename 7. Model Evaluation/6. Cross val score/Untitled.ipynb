```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target_names[iris.target]
```

```python
df.head()
```

```python
X = df.drop(columns=['species'])
y = df['species']
```

# 1. Holdout Method

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,   random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```

# 2. K-Fold Methods

### a. Basic K-Fold

```python
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score

kf = KFold(n_splits=5,shuffle=True,random_state=42)

scores = cross_val_score(model,X ,y ,cv=kf,scoring='accuracy')

print(scores)
print("Mean:", scores.mean())
print("Std:", scores.std())
```

# b. Stratified K-Fold

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import cross_val_score

skf = StratifiedKFold(n_splits=5,shuffle=True,random_state=42)

scores = cross_val_score(model,X,y,cv=skf,scoring='accuracy')

print(scores)
print(scores.mean())
```

### c. Repeated K-Fold

```python
from sklearn.model_selection import RepeatedKFold

rkf = RepeatedKFold(n_splits=5,n_repeats=10,random_state=42)

scores = cross_val_score(model,X,y,cv=rkf)

print("Number of scores:", len(scores))
print("Mean:", scores.mean())
```

### d. Repeated Stratified K-Fold

```python
from sklearn.model_selection import RepeatedStratifiedKFold

rskf = RepeatedStratifiedKFold(n_splits=5,n_repeats=10, random_state=42)

scores = cross_val_score(model,X,y,cv=rskf,scoring='accuracy')

print(scores.mean())
print(scores.std())
```

# 3. Leave out 

### a. Leave-One-Out

```python
from sklearn.model_selection import LeaveOneOut

loo = LeaveOneOut()

scores = cross_val_score(model,X,y,cv=loo)

print("Number of folds:", len(scores))
print("Mean score:", scores.mean())
```

### b. Leave-P-Out

# 4. Random Splitting

### a. ShuffleSplit

```python
from sklearn.model_selection import ShuffleSplit

ss = ShuffleSplit(n_splits=10,test_size=0.20,random_state=42)

scores = cross_val_score(model,X,y,cv=ss)

print(scores)
print(scores.mean())
```

### b. Stratified ShuffleSplit

```python
from sklearn.model_selection import StratifiedShuffleSplit

sss = StratifiedShuffleSplit(n_splits=10,test_size=0.20,random_state=42)

scores = cross_val_score(model,X,y,cv=sss,scoring='accuracy')

print(scores.mean())
```

# 5. Group based

# 7. Nested CV

```python
iris = load_iris()
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target_names[iris.target]
```

```python
X = df.drop(columns=['species'])
y = df['species']
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold

inner_cv = StratifiedKFold(n_splits=3,shuffle=True,random_state=42)

grid = GridSearchCV(LogisticRegression(max_iter=1000),param_grid={'C': [0.01, 0.1, 1, 10]},cv=inner_cv,scoring='accuracy')
```

```python
outer_cv = StratifiedKFold(n_splits=5,shuffle=True,random_state=42)

scores = cross_val_score(grid,X,y,cv=outer_cv,scoring='accuracy')

print(scores)
print(scores.mean())
```
