```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris

from sklearn.preprocessing import StandardScaler

from sklearn.cluster import KMeans

from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score)
```

### LOAD DATA

```python
iris = load_iris()

X = iris.data
y_true = iris.target
```

### SCALE DATA

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

### EVALUATE DIFFERENT K

```python
results = []

for k in range(2, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10)

    labels = model.fit_predict(X_scaled)

    inertia = model.inertia_

    silhouette = silhouette_score(X_scaled,labels)

    db = davies_bouldin_score(X_scaled,labels)

    ch = calinski_harabasz_score(X_scaled,labels)

    results.append({
        "K": k,
        "Inertia": inertia,
        "Silhouette": silhouette,
        "Davies_Bouldin": db,
        "Calinski_Harabasz": ch})

results_df = pd.DataFrame(results)

print(results_df)
```

### Plot All Internal Metrics

```python
fig, axes = plt.subplots(2,2,figsize=(12, 9))

# Inertia
axes[0, 0].plot(
    results_df["K"],
    results_df["Inertia"],
    marker="o")

axes[0, 0].set_title("Inertia / WCSS")
axes[0, 0].set_xlabel("K")

# Silhouette
axes[0, 1].plot(
    results_df["K"],
    results_df["Silhouette"],
    marker="o")

axes[0, 1].set_title("Silhouette Score")
axes[0, 1].set_xlabel("K")


# Davies-Bouldin
axes[1, 0].plot(
    results_df["K"],
    results_df["Davies_Bouldin"],
    marker="o")

axes[1, 0].set_title("Davies-Bouldin Index")
axes[1, 0].set_xlabel("K")

# Calinski-Harabasz
axes[1, 1].plot(
    results_df["K"],
    results_df["Calinski_Harabasz"],
    marker="o")

axes[1, 1].set_title("Calinski-Harabasz Index")
axes[1, 1].set_xlabel("K")

plt.tight_layout()
plt.show()
```

### External Evaluation

```python
from sklearn.metrics import (
    adjusted_rand_score,
    normalized_mutual_info_score,
    homogeneity_score,
    completeness_score,
    v_measure_score)


kmeans = KMeans(n_clusters=3,random_state=42,n_init=10)

labels = kmeans.fit_predict(X_scaled)

ari = adjusted_rand_score(y_true,labels)

nmi = normalized_mutual_info_score(y_true,labels)

homogeneity = homogeneity_score(y_true,labels)

completeness = completeness_score(y_true,labels)

v_measure = v_measure_score(y_true,labels)

print("ARI:", ari)
print("NMI:", nmi)
print("Homogeneity:", homogeneity)
print("Completeness:", completeness)
print("V-measure:", v_measure)
```
