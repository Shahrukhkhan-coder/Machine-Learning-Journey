```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

from sklearn.cluster import (KMeans,AgglomerativeClustering,DBSCAN)
from Cluster_Evaluation_Function import evaluate_clustering
```

```python
iris = load_iris()

X = iris.data
y_true = iris.target
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
models = {"KMeans": KMeans(n_clusters=3, random_state=42, n_init=10),
    "Agglomerative": AgglomerativeClustering(n_clusters=3),
    "DBSCAN": DBSCAN(eps=0.5, min_samples=5)}
```

```python
for name, model in models.items():

    labels = model.fit_predict(X_scaled)

    # Some algorithms can produce
    # only one cluster on unsuitable data.
    if len(set(labels)) > 1:
        scores = evaluate_clustering(X_scaled,labels)

        print("\n",name)
        for metric, score in scores.items():
            print(f"{metric}: {score:.4f}")
```
