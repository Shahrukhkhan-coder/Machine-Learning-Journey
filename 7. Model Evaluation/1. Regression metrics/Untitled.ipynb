```python
ord_col = df.select_dtypes(include='object')
ord_col['Country'].nunique()
```

```python
counts = df['Country'].value_counts()

threshold = 100 

repl = counts[counts<=threshold].index

df['Country'] = df['Country'].replace(repl,'Other')
```

```python
skew_col = ['Study_Hours']

numeric_cols = ['Age','Avg_Daily_Usage_Hours','Daily_Unlocks','Physical_Activity_Hours','Sleep_Hours_Per_Night']

ordinal_cols = ['Academic_Level','Stress_Level']

nominal_cols = ['Gender','Country','Most_Used_Platform','Purpose_Of_Use']

features = skew_col + numeric_cols + ordinal_cols + nominal_cols
x = df[features]
y = df['Mental_Health_Score']
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler, FunctionTransformer, OrdinalEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

study_hours_pipeline = Pipeline(steps=[
    ('log_transform', FunctionTransformer(np.log1p)),
    ('scaler', StandardScaler())
])

preprocessing = ColumnTransformer(transformers=[
    ('study_hours', study_hours_pipeline, skew_col),
    ('numer',StandardScaler(),numeric_cols),
    ('ordinal',OrdinalEncoder(categories=[['High School', 'Undergraduate', 'Graduate'],['Low', 'Medium', 'High', 'Very High']]),ordinal_cols),
    ('nominal',OneHotEncoder(drop='first',sparse_output=False,dtype=np.int32),nominal_cols)
])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

rf_pipeline = Pipeline(steps=[
                       ('preprocessing',preprocessing),
                       ('regression',RandomForestRegressor())
])
```

```python
rf_pipeline.fit(x_train, y_train)

rf_pred_test = rf_pipeline.predict(x_test)
rf_pred_train = rf_pipeline.predict(x_train)

r2_test = r2_score(y_test,rf_pred_test)
r2_train = r2_score(y_train,rf_pred_train)
mae_score = mean_absolute_error(y_test, rf_pred_test)

print(f"Accuracy of Training {r2_train}")
print(f"Accuracy of Testing {r2_test}")
print(f'MAE {mae_score}')
```

```python
testing_accuracy = mean_squared_error(y_test,rf_pred_test)
traning_accuracy = mean_squared_error(y_train,rf_pred_train)
print(f"MSE of Training {traning_accuracy}")
print(f"MSE of Testing {testing_accuracy}")
```

```python
testing_accuracy = np.sqrt(mean_squared_error(y_test,rf_pred_test))
traning_accuracy = np.sqrt(mean_squared_error(y_train,rf_pred_train))
print(f"RMSE of Training {traning_accuracy}")
print(f"RMSE of Testing {testing_accuracy}")
```
